```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import zipfile
from io import BytesIO
import re
```

```python
# Step1: Read Data from Zip file
```

```python
# Path to your ZIP file
zip_path = "C:/Users/AHSAN RAZA/Desktop/Projects for Machine Learning/Study Materials/Project - Analyzing Naming Trends/names.zip"

# List to store yearly DataFrames
dfs = []

# Open the ZIP file
with zipfile.ZipFile(zip_path, 'r') as z:

    # Loop through all files in the ZIP
    for file_name in z.namelist():

        # Keep only files with pattern "yobYYYY.txt"
        match = re.match(r"yob(\d{4})\.txt$", file_name)

        # Only accept files from 1880 onwards
        if match and int(match.group(1)) >= 1880:

            # Extract year from filename
            year = int(match.group(1))

            # Read the file into a DataFrame
            df = pd.read_csv(
                BytesIO(z.read(file_name)),
                encoding="utf-8",
                engine="python",
                header=None,
                names=["name", "sex", "count"]
            )

            # Add year column
            df["year"] = year

            # Append to list
            dfs.append(df)

# Combine all years into one DataFrame
names_df = pd.concat(dfs, ignore_index=True)

# Show first 5 rows
print(names_df.head())
```

```python
print(names_df)
```

```python
# Step2: Visualize male and female births in a particular year
```

```python
# Choose a year to visualize
selected_year = 1990

# Filter data for the chosen year
year_data = names_df[names_df["year"] == selected_year]

# Group by sex and sum the counts
gender_count = year_data.groupby("sex")["count"].sum()

# Plot male vs female births
gender_count.plot(kind="bar")
plt.title(f"Male vs Female births in {selected_year}")
plt.xlabel("Sex")
plt.ylabel("Number of births")
plt.show()
```

```python
# Step3: Find top 100 baby names in a year
```

```python
# Filter for the selected year
year_data = names_df[names_df["year"] == selected_year]

# Sort by birth count in descending order
top_100 = year_data.sort_values(by="count", ascending=False).head(100)

# Show top 100
print(top_100.head())
```

```python
# Step4: Find popular baby names (grouped by name)
```

```python
# Take top 100 entries
top_100 = year_data.sort_values(by="count", ascending=False).head(100)

# Group by name to get total counts
popular_names = top_100.groupby("name")["count"].sum().sort_values(ascending=False)

# Show popular names
print(popular_names.head(20))
```

```python
# Step5: Save results to CSV
```

```python
popular_names.to_csv("popular_names_1990.csv")
```
