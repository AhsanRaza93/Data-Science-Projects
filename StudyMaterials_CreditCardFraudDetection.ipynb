<h2>Credit Card Fraud Detection</h2>

Problem Statement:

1. A credit card is a small thin plastic or fiber card that incorporates
information about the person such as a picture or signature and the
person’s name on it to charge purchases and services to his linked
account. Charges are debited regularly. Nowadays, card data is read by
ATMs, swiping machines, store readers, banks and online transactions.

2. Each card has a unique card number which is very important. Its security
mainly relies on the physical security of the card and also the privacy of the
credit card number. There is a rapid growth in credit card transactions
which has led to substantial growth in scam cases.

3. Credit card fraud is expanding heavily because fraud financial loss is
increasing drastically. Multiple data mining and statistical techniques are
used to catch fraud. Therefore the detection of fraud using efficient and
secured methods are very important.

3. From the dataset, calculate the number of genuine transactions, number of

model.

7. Now use a decision tree and random forest model for training on top of the
train set.

8. Compare the predictions of both models using predict().

9. Compare the accuracy of both models using score().

10. Check the performance matrix of both models and compare which

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
credit_fraud = pd.read_csv("C:/Users/AHSAN RAZA/Desktop/creditcard.csv.crdownload")
print(credit_fraud)
```

```python
# Missing value analysis

missing_values = credit_fraud.isnull().sum()
print(missing_values)

credit_fraud = credit_fraud.dropna()
```

```python
credit_fraud.isnull().sum()
```

```python
# Calsulating no. of genuine transaction, fraud transaction and percentage of fraud transaction
genuine_transaction = credit_fraud[credit_fraud['Class'] == 0].shape[0]
fraud_transaction = credit_fraud[credit_fraud['Class'] == 1].shape[0]
fraud_transaction_percentage = (fraud_transaction /credit_fraud.shape[0])*100

print('Number of Genuine transactions:\n', genuine_transaction)
print('Number of Fraud transactions:\n', fraud_transaction)
print('Percentage of Fraud transactions:\n', fraud_transaction_percentage)
```

```python
# Visualizing genuine and fraud transactions

counts = credit_fraud['Class'].value_counts()

plt.bar(['Genuine','Fraud'],counts.values, color=['Green','Red'])
plt.xlabel('Transaction Type')
plt.ylabel('Count')
plt.title('Genuine vs Fraudulent Transactions')
plt.show()
```

```python
# Normalize the Amount column

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
credit_fraud['Normalized Amount'] = scaler.fit_transform(credit_fraud[['Amount']])

credit_fraud = credit_fraud.drop(['Amount'], axis=1)
```

```python
# Spliting the dataset in train-test set

from sklearn.model_selection import train_test_split

x = credit_fraud.drop('Class', axis=1)
y = credit_fraud['Class']

x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.3,random_state=42,stratify=y)
```

```python
# Decision Tree and Random Forest model 

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

dt = DecisionTreeClassifier(random_state=42)
dt.fit(x_train, y_train)

rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(x_train, y_train)
```

```python
# Comparing predictions of both the models

dt_preds = dt.predict(x_test)
rf_preds = rf.predict(x_test)
```

```python
# Comparing accuracy using score()

dt_accuracy = dt.score(x_test, y_test)
rf_accuracy = dt.score(x_test, y_test)

print('Decision Tree Accuracy:\n', dt_accuracy)
print('Random Forest Accuracy:\n', rf_accuracy)
```

```python
# Performace matrix of both the models

from sklearn.metrics import confusion_matrix, classification_report

print('Decision Tree confusion matrix:\n')
print(confusion_matrix(y_test, dt_preds))
print(classification_report(y_test, dt_preds))

print('Random Forest confusion matrix:\n')
print(confusion_matrix(y_test, dt_preds))
print(classification_report(y_test, dt_preds))
```

```python
# Performance comparision

from sklearn.metrics import precision_score, recall_score, f1_score, roc_auc_score

# Decision Tree metrics
dt_precision = precision_score(y_test, dt_preds)
dt_recall = recall_score(y_test, dt_preds)
dt_f1 = f1_score(y_test, dt_preds)
dt_auc = roc_auc_score(y_test, dt_preds)

# Random Forest metrics
rf_precision = precision_score(y_test, rf_preds)
rf_recall = recall_score(y_test, rf_preds)
rf_f1 = f1_score(y_test, rf_preds)
rf_auc = roc_auc_score(y_test, rf_preds)

# Display results
print("Decision Tree Performance:")
print("Precision:", dt_precision)
print("Recall:", dt_recall)
print("F1 Score:", dt_f1)
print("ROC-AUC:", dt_auc)

print("\nRandom Forest Performance:")
print("Precision:", rf_precision)
print("Recall:", rf_recall)
print("F1 Score:", rf_f1)
print("ROC-AUC:", rf_auc)
```

```python
# Tabular comparison: This makes it very clear which model is better.

performance_df = pd.DataFrame({
    'Model': ['Decision Tree', 'Random Forest'],
    'Precision': [dt_precision, rf_precision],
    'Recall': [dt_recall, rf_recall],
    'F1 Score': [dt_f1, rf_f1],
    'ROC-AUC': [dt_auc, rf_auc]
})

print(performance_df)
```
