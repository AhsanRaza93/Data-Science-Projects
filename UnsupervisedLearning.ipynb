```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
customers = pd.read_csv("C:/Users/AHSAN RAZA/Desktop/All My Files/Data Science/Intellipaat/PYTHON/Python2/Study Materials/Module 9 - Unsupervised Learning/Assignment 1/customers.csv")
print(customers)
```

```python
customers.isnull().sum()
```

```python
x = customers[["Age"]]
y = customers[["Spending Score (1-100)"]]
```

```python
plt.scatter(x,y,color="blue",marker="o",label="Data Points")
plt.xlabel('Age')
plt.ylabel('Spending Score (1-100)')
plt.title('Age VS Spending Score')
plt.legend()
plt.show()
```

```python
from sklearn.cluster import KMeans
```

```python
x = customers[['Annual Income (k$)', 'Spending Score (1-100)']]
kmeans = KMeans(n_clusters=5, random_state=42)
customers['Cluster'] = kmeans.fit_predict(x)
plt.figure(figsize=(10, 6))
colors = ['red', 'blue', 'green', 'orange', 'purple']
for cluster in range(5):
    cluster_data = customers[customers['Cluster'] == cluster]
    plt.scatter(cluster_data['Annual Income (k$)'], cluster_data['Spending Score (1-100)'],label=f'Cluster {cluster}', color=colors[cluster], s=60)
centers = kmeans.cluster_centers_
plt.scatter(centers[:, 0], centers[:, 1], c='black', s=200, marker='X', label='Centroids')
plt.title('Customer Segments Based on Income and Spending Score')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
x = customers1[['Annual Income (k$)', 'Spending Score (1-100)']]
inertias = []
K_range = range(1, 11)
for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(x)
    inertias.append(kmeans.inertia_)
plt.figure(figsize=(8, 5))
plt.plot(K_range, inertias, marker='o', linestyle='--')
plt.title('Elbow Method for Optimal Number of Clusters')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Inertia (Within-Cluster Sum of Squares)')
plt.xticks(K_range)
plt.grid(True)
plt.show()
```

```python
x = customers1[['Annual Income (k$)', 'Spending Score (1-100)']]
kmeans = KMeans(n_clusters=5, random_state=42)
customers1['Cluster'] = kmeans.fit_predict(x)
plt.figure(figsize=(10, 6))
colors = ['red', 'blue', 'green', 'orange', 'purple']
for cluster in range(5):
    cluster_data = customers1[customers1['Cluster'] == cluster]
    plt.scatter(cluster_data['Annual Income (k$)'], cluster_data['Spending Score (1-100)'],label=f'Cluster {cluster}', color=colors[cluster], s=60)
centers = kmeans.cluster_centers_
plt.scatter(centers[:, 0], centers[:, 1], c='black', s=200, marker='X', label='Centroids')
plt.title('Customer Segments Based on Income and Spending Score')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
data = {
    'milk':     [1, 1, 0, 1, 0],
    'bread':    [1, 1, 1, 1, 0],
    'butter':   [0, 1, 1, 0, 0],
    'jam':      [0, 0, 0, 1, 1],
    'eggs':     [1, 0, 1, 0, 1]
}
df = pd.DataFrame(data)
frequent_itemsets = apriori(df, min_support=0.4, use_colnames=True)
rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1.0)
print("📦 Frequent Itemsets:")
print(frequent_itemsets)
print("\n🔗 Association Rules:")
print(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']])
```

```python
reader = Reader(rating_scale=(ratings.rating.min(), ratings.rating.max()))
data = Dataset.load_from_df(ratings[['userId', 'movieId', 'rating']], reader)
trainset = data.build_full_trainset()
svd = SVD()
svd.fit(trainset)
user_id = 5
user_rated_movies = ratings[ratings['userId'] == user_id]['movieId'].tolist()
all_movie_ids = movies['movieId'].unique()
unrated_movies = [movie_id for movie_id in all_movie_ids if movie_id not in user_rated_movies]
predictions = []
for movie_id in unrated_movies:
    pred = svd.predict(user_id, movie_id)
    predictions.append((movie_id, pred.est))
pred_df = pd.DataFrame(predictions, columns=['movieId', 'predicted_rating'])
pred_df = pred_df.merge(movies, on='movieId', how='left')
top_10 = pred_df.sort_values(by='predicted_rating', ascending=False).head(10)
print(top_10[['movieId', 'title', 'predicted_rating']])
```

```python
import statsmodels.api as sm
mtcars = sm.datasets.get_rdataset("mtcars", "datasets").data
car_features = mtcars[['mpg', 'disp', 'hp']]
print(car_features)
```

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
car_features_scaled = scaler.fit_transform(car_features)
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(car_features_scaled)
car_features['cluster'] = kmeans.labels_
print(car_features)
```

```python
for i in sorted(car_features['cluster'].unique()):
    print(f"\nCluster {i}:")
    print(car_features[car_features['cluster'] == i])
```

```python
kmeans = KMeans(n_clusters=5, random_state=42)
kmeans.fit(car_features_scaled)
car_features['cluster'] = kmeans.labels_
print(car_features)
```

```python
for i in sorted(car_features['cluster'].unique()):
    print(f"\nCluster {i}:")
    print(car_features[car_features['cluster'] == i])
```
