```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
churn2 = pd.read_csv("C:/Users/AHSAN RAZA/Desktop/Projects for Machine Learning/Study Materials/Project - Customer Churn/customer_churn.csv")
print(churn2)
```

```python
# A.a) Extract the 5th column & store it in ‘customer_5’
customer_5 = churn2.iloc[:,4]
print(customer_5)
```

```python
# A.b.) Extract the 15th column & store it in ‘customer_15’
customer_15 = churn2.iloc[:,14]
print(customer_15)
```

```python
# A.c.) Extract all the male senior citizens whose Payment Method is Electronic check & store the result in ‘senior_male_electronic’
senior_male_electronic = churn2[(churn2['gender']=='Male')&(churn2['SeniorCitizen']==1)&(churn2['PaymentMethod']=='Electronic check')]
print(senior_male_electronic)
```

```python
# A.d.) Extract all those customers whose tenure is greater than 70 months or their Monthly charges is more than 100$ & store the result in ‘customer_total_tenure’
customer_total_tenure = churn2[(churn2['tenure']>70)|(churn2['MonthlyCharges']>100)]
print(customer_total_tenure)
```

```python
# A.e.) Extract all the customers whose Contract is of two years, payment method is Mailed check & the value of Churn is ‘Yes’ & store the result in ‘two_mail_yes’
two_mail_yes = churn2[(churn2['Contract']=='Two year')&(churn2['PaymentMethod']=='Mailed check')&(churn2['Churn']=='Yes')]
print(two_mail_yes)
```

```python
# A.f.) Extract 333 random records from the customer_churn dataframe& store the result in ‘customer_333’
customer_333 = churn2.sample(n=333)
print(customer_333) 
```

```python
# A.g.) Get the count of different levels from the ‘Churn’ column
churn2['Churn'].value_counts()
```

```python
# B.a) Build a bar-plot for the ’InternetService’ column:
internet_service_counts = churn2['InternetService'].value_counts()
plt.bar(internet_service_counts.index, internet_service_counts.values, color='orange')
plt.xlabel('Categories of Internet Service')
plt.ylabel('Count of Categories')
plt.title('Distribution of Internet Service')
plt.show()
```

```python
# B.b) Build a histogram for the ‘tenure’ column:
sns.histplot(churn2['tenure'], bins=30, color='green')
plt.title('Distribution of tenure')
plt.show()
```

```python
# B.c.) Build a scatter-plot between ‘MonthlyCharges’ & ‘tenure’. Map ‘MonthlyCharges’ to the y-axis & ‘tenure’ to the ‘x-axis’:
plt.scatter(churn2['tenure'],churn2['MonthlyCharges'], color='brown', s=10)
plt.xlabel('Tenure of customer')
plt.ylabel('Monthly Charges of customer')
plt.title('Tenure vs Monthly Charges')
plt.show()
```

```python
# B.d.) Build a box-plot between ‘tenure’ & ‘Contract’. Map ‘tenure’ on the y-axis & ‘Contract’ on the x-axis.
sns.boxplot(x='Contract', y='tenure', data=churn2)
plt.xlabel('Contract')
plt.ylabel('tenure')
plt.title('Tenure vs Contract')
plt.show()
```

```python
# C.a) Linear Regression:

x = churn2[['tenure']]
y = churn2[['MonthlyCharges']]

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=42)
le = LinearRegression()
le.fit(x_train, y_train)

y_pred = le.predict(x_test)

mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

error = y_test.values - y_pred
rmse = np.sqrt(mse)

print("Mean Squared Error", mse)
print("Root Mean Squared Error", rmse)
print("R2 Score", r2)
```

```python
# D.a.) Logistic Regression

x = churn2[['MonthlyCharges']]
y = churn2['Churn']

# If Churn is categorical (Yes/No), convert it first:
y = y.map({'Yes': 1, 'No': 0})

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, accuracy_score

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.35, random_state=42)

log_model = LogisticRegression()
log_model.fit(x_train, y_train)

y_pred = log_model.predict(x_test)

cm = confusion_matrix(y_test, y_pred)
acc = accuracy_score(y_test, y_pred)

print("Confusion Matrix:\n", cm)
print("Accuracy Score:", acc)
```

```python
# D.b.) Logistic Regression

x = churn2[['tenure','MonthlyCharges']]
y = churn2['Churn']

# If Churn is categorical (Yes/No), convert it first:
y = y.map({'Yes': 1, 'No': 0})

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, accuracy_score

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

log_model = LogisticRegression(max_iter=1000)
log_model.fit(x_train, y_train)

y_pred = log_model.predict(x_test)

cm = confusion_matrix(y_test, y_pred)
acc = accuracy_score(y_test, y_pred)

print("Confusion Matrix:\n", cm)
print("Accuracy Score:", acc)
```

```python
# E.a.) Decision Tree

from sklearn.tree import DecisionTreeClassifier

x = churn2[['tenure']]
y = churn2['Churn']

y = y.map({'Yes':1, 'No':0})

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(x_train, y_train)

y_pred = dt_model.predict(x_test)

cm = confusion_matrix(y_test, y_pred)
acc = accuracy_score(y_test, y_pred)

print("Confusion Matrix:\n", cm)
print("Accuracy Score:", acc)
```

```python
# F.a.) Random Forest

from sklearn.ensemble import RandomForestClassifier

x = churn2[['tenure','MonthlyCharges']]
y = churn2['Churn']

y = y.map({'Yes':1, 'No':0})
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.3, random_state=42)

rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(x_train, y_train)

y_pred = rf_model.predict(x_test)

cm = confusion_matrix(y_test, y_pred)
acc = accuracy_score(y_test, y_pred)

print("Confusion Matrix:\n", cm)
print("Accuracy_score:", acc)
```
